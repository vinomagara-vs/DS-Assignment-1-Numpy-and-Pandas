# numpy

```python
import numpy as np
```

```python
# 1D numpy

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(f"1D array: {temperatures_w1} and its type is {type(temperatures_w1)}")
```

```python
# shape
print(f"Array dimension is {temperatures_w1.ndim}")

#data type of the values in the array
print(f"data type is {temperatures_w1.dtype}")

#no of elements
print(f"no of elements of temperatures_w1 is {temperatures_w1.size}")
print(f"shapes of temperatures_w1 is {temperatures_w1.shape}")
```

```python
# array operations

Fahrenheit_w1=np.array([])

for i in temperatures_w1:
  Fahrenheit = (temperatures_w1 * 9/5) + 32
  Fahrenheit_w1 = np.append(Fahrenheit_w1,Fahrenheit)
print(f"Temperatures in Fahrenheit: {Fahrenheit}")

# minimum temp of the week

print(f"Minimum temperature: {np.min(Fahrenheit_w1)}")

# maximum temp of the week
print(f"Maximum temperature: {np.max(Fahrenheit_w1)}")

# mean temp of the week

print(f"Mean temperature: {np.mean(Fahrenheit_w1)}")
```

```python
# Array slicing and indexing

#1D array: [22.5 25.3 20.8 23.4 26.1 24.8 21.9]

"""
print(f"Temperature of 1st 3 days: {temperatures_w1[range(0,3)]}")
print(f"Temperature of last 2 days: {temperatures_w1[-2:]}")
print(f"Temperature of last 2 days: {temperatures_w1[range(2,5)]}")"""

print(f"Temperature of 1st 3 days: {temperatures_w1[3:]}")
print(f"Temperature of last 2 days: {temperatures_w1[-2:]}")
print(f"Temperature of last 2 days: {temperatures_w1[2:5]}")
```

```python
# 2 D array

temperatures=np.array( [[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] , [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
print(f"Temperatures: {temperatures}")
```

```python
# inspect the shape and elements

print(f"Array dimension is {temperatures.ndim}")
print(f"data type of the temperatures is {temperatures.dtype}")
print(f"no of elements in temperatures is {temperatures.size}")
print(f"shape of temperatures array is {temperatures.shape}")
```

```python
# indexing and slicing for 2d array

print(f"First week of temperature is: {temperatures[0]}")
print(f"Second week of temperature is: {temperatures[1]}")
print(f"weekend in 1st week of temperature is: {temperatures[0][-2:]}")
print(f"weekend in 2nd week of temperature is: {temperatures[1][-2:]}")
```

# Pandas Data Frame


```python
import pandas as pd
```

```python
# creating pandas dataframe

transactions = {'TransactionID' : [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
                'ProductCategory' : ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing','Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
                'Region' : ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South','North'],
                'Amount' : [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
                }
print(transactions)

# define the transaction to dataframe

transactions_df = pd.DataFrame(transactions)
print(transactions_df)
```

```python
# Data exploration

print("Basic info for dataframe")
print(transactions_df.info())
```

```python
print(transactions_df.head()) # diplay the 1st 5 records
```

```python
print(transactions_df.tail()) # diplay the last 5 records
```

```python
print(transactions_df.describe)
```

```python
# Display product & amount columns in df

transactions_df[['ProductCategory','Amount']]
```

```python
# retrieve last 3 columns

print(f"Retrieve last 3 columns:\n {transactions_df.iloc[0:,-3:]}")
```

```python

# fetch the data after filter condition

print(f"filter condition:\n {transactions_df.loc[(transactions_df['Region']=='North') & (transactions_df['Amount'] > 200)]}")
```

```python
#get the counts of product after group by
print(f"Product category aggr:\n {transactions_df['ProductCategory'].value_counts()}")
```

```python
#Unique values of region
print(f"Unique Region Values:\n {transactions_df['Region'].nunique()}")
```

```python
# mean value for all region

print(f"Mean value for all region:\n {transactions_df.groupby('Region')['Amount'].mean()}")
```

```python
#Manipulating the data frame

#update the amount value for transaction id is 102
transactions_df.loc[transactions_df['TransactionID'] == 102 , 'Amount']=165
print(f"Update the amount value for 102 id: {transactions_df}")
```

```python
# add the discount column with 10%

transactions_df['Discount']=transactions_df['Amount']*0.10
print(f"After added the discount column:\n,{transactions_df}")
```

```python
# remove the row of transaciton id 109

transactions_df.drop(transactions_df[transactions_df['TransactionID']==109].index,inplace=True)
print(f"After remove the row of transaciton id 109:\n,{transactions_df}")
```

```python
#remove the discount column

transactions_df.drop('Discount',axis=1,inplace=True)
print(f"After remove the discount column:\n,{transactions_df}")
```
